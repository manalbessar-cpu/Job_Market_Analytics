```python
df["salary_min"].median()
df["salary_min"] = df["salary_min"].fillna(df["salary_min"].median())
df.isnull().sum()
```

```python
df["posted_date"] = pd.to_datetime(df["posted_date"])
df["posted_date"].dtype
```

```python
df["employment_type"] = df["employment_type"].replace({
    "Full time": "Full-time"
})
df["employment_type"].value_counts()
```

```python
df["salary_avg"] = (df["salary_min"] + df["salary_max"]) / 2
df[["salary_min", "salary_max", "salary_avg"]].head()
```

```python
df["posted_year"] = df["posted_date"].dt.year
df["posted_month"] = df["posted_date"].dt.month
df[["posted_date", "posted_year"]].head()
df[["posted_date", "posted_month"]].head()
```
