```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

df = pd.read_csv("C:/Users/manal/OneDrive/Desktop/Job Market Analytics/data/job_market_cleaned.csv")
```

```python
top_jobs = df["job_title"].value_counts().head(10)

top_jobs
```

```python
plt.figure(figsize=(10, 6))

top_jobs.plot(
    kind="bar",
    color="#FF69B4",     
    edgecolor="black"
)

plt.title("Top 10 Most Demanded Job Titles", fontsize=16, fontweight="bold")
plt.xlabel("Job Title", fontsize=12)
plt.ylabel("Number of Job Offers", fontsize=12)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

plt.show()
```

```python
top_companies = df["company_name"].value_counts().head(10)

plt.figure(figsize=(10,6))

top_companies.plot(
    kind="bar",
    color="#FF69B4",     
    edgecolor="black"
)

plt.title("Top 10 Companies Hiring", fontsize=16, fontweight="bold")
plt.xlabel("Company", fontsize=12)
plt.ylabel("Number of Job Offers", fontsize=12)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.hist(
    df["salary_avg"].dropna(),
    bins=30,
    color="#FF69B4",      
    edgecolor="black"
)

plt.title("Salary Distribution", fontsize=16, fontweight="bold")
plt.xlabel("Average Salary", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
remote = df["remote_type"].value_counts()

plt.figure(figsize=(7,7))

remote.plot(
    kind="pie",
    autopct="%1.1f%%",
    colors=["#FF69B4", "#FFC0CB", "#FFB6C1"], 
    wedgeprops={"edgecolor": "black", "linewidth": 1}
)

plt.title("Remote Work Distribution", fontsize=16, fontweight="bold")
plt.ylabel("")

plt.tight_layout()
plt.show()
```

```python
skills = (
    df["skills"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

top_skills = skills.value_counts().head(10)

plt.figure(figsize=(10,6))

top_skills.plot(
    kind="bar",
    color="#FF69B4",      
    edgecolor="black"
)

plt.title("Top 10 Most Required Skills", fontsize=16, fontweight="bold")
plt.xlabel("Skill", fontsize=12)
plt.ylabel("Count", fontsize=12)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
top_countries = df["country"].value_counts().head(10)

plt.figure(figsize=(10,6))

top_countries.plot(
    kind="bar",
    color="#FF69B4",     
    edgecolor="black"
)

plt.title("Top 10 Countries", fontsize=16, fontweight="bold")
plt.xlabel("Country", fontsize=12)
plt.ylabel("Number of Jobs", fontsize=12)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
