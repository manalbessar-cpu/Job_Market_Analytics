```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/manal/OneDrive/Desktop/Job Market Analytics/data/job_market_cleaned.csv")
```

```python
top_jobs = df["job_title"].value_counts().head(10)

top_jobs
```

```python
plt.figure(figsize=(10, 6))

top_jobs.plot(kind="bar")

plt.title("Top 10 Most Demanded Job Titles")
plt.xlabel("Job Title")
plt.ylabel("Number of Job Offers")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```
